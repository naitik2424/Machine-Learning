```python
import pickle
import numpy as np
```

```python
ohe_sex=pickle.load(open('model/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('model/ohe_embarked.pkl','rb'))
clf= pickle.load(open('model/clf.pkl','rb'))
```

```python
test_input=np.array([2,'male',31.0,0,0,10.5,'S'],dtype=object).reshape(1,7)
```

```python
test_input
```

```python
import pandas as pd

test_input_sex = ohe_sex.transform(
    pd.DataFrame(test_input[:, 1].reshape(-1, 1), columns=['Sex'])
)
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(
    pd.DataFrame(
        test_input[:, -1].reshape(1, 1),
        columns=['Embarked']
    )
)
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]],
test_input_age,test_input_sex,test_input_embarked),axis=1)
```

```python
test_input_transformed = np.concatenate(
    (
        test_input[:, [0, 3, 4, 5]],
        test_input_age,
        test_input_sex,
        test_input_embarked
    ),
    axis=1
)
```
